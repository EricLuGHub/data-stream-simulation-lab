```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/prices_round_2_day_-2.csv", sep=";")
df.head()
```

```python
product = "ASH_COATED_OSMIUM"

df_p = df[df["product"] == product]
df_p = df_p[df_p["mid_price"] > 0]

plt.plot(df_p["timestamp"], df_p["mid_price"])
plt.title(product)
plt.show()
```
